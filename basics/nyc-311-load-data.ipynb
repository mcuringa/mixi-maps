```python
import pandas as pd
# change this variable to load more of fewer records
# in python you can use _ to make large numbers more readable (optional)
num_records = 1_000_000
url = f"https://data.cityofnewyork.us/resource/erm2-nwe9.csv?$limit={num_records}"
df = pd.read_csv(url)
df.head()
```

Use value_counts() to explore data
----------------------------------

```python

# this will show us all of the unique `complaint_type` values and
# how many times each one appears in the dataset
df["complaint_type"].value_counts()
```

```python
# create a new dataframe with the to 20 complaints
complaints = df["complaint_type"].value_counts().head(20)
complaints = pd.DataFrame(complaints).reset_index()
complaints
```

```python
# get just the parking complaints
parking = df[df.complaint_type == "Illegal Parking"]
# we can look at the "descriptor" column for more details
parking["descriptor"].value_counts()
```

```python
# save the dataframe to a CSV file so we don't need to load it from Open Data again
df.to_csv("nyc-311.csv", index=False)
```
