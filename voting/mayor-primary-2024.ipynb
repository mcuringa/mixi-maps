```python
# autoreload
%load_ext autoreload
%autoreload 2
import pandas as pd
import geopandas as gpd
from miximaps import ui, nyc
import folium
```

```python
# election results for 2022 NYC Mayor
url = "https://vote.nyc/sites/default/files/pdf/election_results/2021/20211102General%20Election/00000200000Citywide%20Mayor%20Citywide%20EDLevel.csv"
raw = pd.read_csv(url, header=None)
```

```python
# confirm how many columns we need to get the actual headers
cols = raw.columns[:11]
raw[cols].head(2)
```

```python
df  = raw.copy()
# read the col names from the first row
cols = raw.iloc[0][raw.columns[:11]].to_list()
df.drop(columns=df.columns[:len(cols)], inplace=True)
df.columns = cols
# df["Tally"] = pd.to_numeric(df["Tally"].str.replace( ",", ""), errors="coerce").astype("Int64")
df.Tally.mean()
```

```python
# find out how the candidates are
all = set(df["Unit Name"].unique())
special  = {'Public Counter',
 'Manually Counted Emergency',
 'Absentee / Military',
 'Affidavit',
 'Scattered'}
candidates = all - special
```

```python
df[["Unit Name", "Tally"]].groupby("Unit Name").sum().reset_index().sort_values("Tally", ascending=False)
```

```python
df["Candidate"] = df["Unit Name"].map({
    "Donald J. Trump / JD Vance (Republican)": "Trump",
    "Donald J. Trump / JD Vance (Conservative)": "Trump",
    "Kamala D. Harris / Tim Walz (Democratic)": "Harris",
    "Kamala D. Harris / Tim Walz (Working Families)": "Harris",
})


df = df[df["Candidate"].notna()].copy()
df = df[['AD', 'ED', 'County', 'Candidate', 'Tally']]
df
```

```python

results = df.groupby(['AD', 'ED', 'County', 'Candidate'], as_index=False).sum()
results = results.pivot(index=["AD", "ED", "County"], columns="Candidate", values="Tally").reset_index().rename_axis(columns=None)
results["Total"] = results.Harris + results.Trump
results["Harris %"] = results.Harris / results.Total
results["Trump %"] = results.Trump / results.Total
results["Winner"] = results[["Harris", "Trump"]].idxmax(axis=1)
results
```

```python
url = "https://data.cityofnewyork.us/resource/wwxk-38u4.geojson?$limit=20000"
ed = gpd.read_file(url)
ed.columns
```

```python
results["elect_dist"] = results.apply(lambda x: f"{x.AD}{x.ED:03d}", axis=1)
results
```

```python

map_data = results.merge(ed[["elect_dist", "geometry"]], on="elect_dist", how="left")
map_data = map_data[map_data.Total > 0]
map_data = gpd.GeoDataFrame(map_data, crs=ed.crs)
"No reported tally:", len(results) - len(map_data)  # missing some districts, ~90 with no election total
```

```python
m = ui.base_map(map_data, name="Election Results")
map_data["tooltip"] = map_data.apply( ui.popup(cols=["Harris %", "Trump %"], title=False), axis=1)
map_data["popup"] = map_data.apply(ui.popup(cols=["Winner", "----", "AD", "ED", "Trump", "Harris"]), axis=1)

m = map_data[map_data["Winner"] == "Trump"].explore(m=m, column="Trump %", cmap="Reds", 
    name="Trump", tooltip="tooltip", popup="popup", popup_kwds={"labels":False,}, tooltip_kwds={"labels":False, "style":"width: 100px"}, legend=False)

m = map_data[map_data["Winner"] == "Harris"].explore(m=m, column="Harris %", cmap="Blues",
    name="Harris", tooltip="tooltip", popup="popup", popup_kwds={"labels":False,}, tooltip_kwds={"labels":False, "style":"width: 100px"}, legend=False)
m
```
