```python
gc = geonamescache.GeonamesCache()
counties = gc.get_us_counties()
county_mapper = dict([(c["fips"], c["name"]) for c in counties])


table = "B25044"
fields = mc.table_vars(table)
vars = list(fields.keys()) # Census wants a real list

county_fips = {
    "36": ["005", "047", "061", "081", "085", "119", "087", "059", "103"],  # NY
    "34": ["003", "013", "017"],  # NJ: Bergen, Essex, Hudson
    "09": ["001"], # CT: Fairfield
}
pairs = [(s, c) for s, counties in county_fips.items() for c in counties]

results = []
for state, counties in pairs:
    r = c.acs5.state_county_tract(vars, state, counties, Census.ALL)
    results.extend(r)

df = pd.DataFrame(results)
df.rename(columns=fields, inplace=True)

# drop rows with no data
df.dropna(inplace=True)
df
```

```python
# give nicer names to counties and states



df["county_name"] = df.apply(mc.county_mapper(), axis=1)
df["statefp"] = df["state"]
df["state"] = df.statefp.apply(mc.lookup_state)
df.columns
```

```python
df
```

```python
cars = df.copy()
cars["no_car"] = cars["renter_occupied_no_vehicle_available"] + cars["owner_occupied_no_vehicle_available"]
cars["has_car"] = cars["total"] - cars["no_car"]
cars["car_owner_pct"] = cars["has_car"] / cars["total"]
cars
```

```python
# use pygris to look up the tract geometries then merge with our data

# get the tracts for all of the sates in our data set
state_fips = cars.statefp.unique().tolist()
tracts = pd.concat([pygris.tracts(state=s, year=2023) for s in state_fips])
cars = cars.merge(tracts[["GEOIDFQ", "geometry"]], left_on="geography", right_on="GEOIDFQ")
cars = gpd.GeoDataFrame(cars, geometry="geometry")
```
