Changing Rent
=============
This notebook uses a choropleth to show changing rents over time (2010-2024)
in NYC census tracts.

**Concepts:**
- using `for` loops to pull multiple years of census data
- using `dict` data structure to organize saved data
- adding multiple layers to a map
- using layer control to toggle between layers
- saving a map as a stand-alone web page

```python

from miximaps import nyc, ui
import folium
from census import Census
import os

api_key = os.environ["CENSUS_API_KEY"]
```

```python
years = range(2010, 2024)
table = "B25064"  # median rent
data = {}
for year in years:
    c = Census(api_key, year=year)
    df = nyc.get_tracts(c, table, year=year, region="city")
    df["year"] = year
    data[year] = df
```

```python
# cols = [c for c in df.columns for df in data.values()]
# cols = set(cols)
cols = [
        'year',
        'borough',
        'tract',
        'median_gross_rent',
        'county',
        'state',
        'countyfp',
        'statefp',
        'geographic_area_name',
        'geography',
        'geometry',
        ]
for year, df in data.items():
    data[year] = df[cols]

data[2010].sample(10)
```

```python
m = ui.base_map(data[2023])
for year in data.keys():
    x = data[year]
    x = x[x.median_gross_rent > 0]
    x.explore(m=m, column="median_gross_rent", cmap="Purples", legend=False,
              name=str(year), style_kwds={"fillOpacity": 1})

folium.LayerControl().add_to(m)
m
```

```python
m.save("rent.html")
```
